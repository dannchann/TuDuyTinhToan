W3A1 : Nhập vào một số nguyên dương có 3 chữ số. Tìm và in ra màn hình “SỐ
NGUYÊN” đảo ngược của n. Không dùng kiểu chuỗi. Chú ý: kiểm tra chỉ xử
lý các số nguyên dương có đúng 3 chữ số. Các số khác không xử lý.

```python
n = int(input())
if n < 100 or n > 999:
    print("Sai dinh dang yeu cau")
else: 
    hang_tram = n//100 
    hang_chuc = (n//10)%10
    hang_don_vi = n%10
print(hang_don_vi*100 + hang_chuc*10 + hang_tram)
```

W3A3 : Kiểm tra xem một số có phải là lũy thừa của 2 hay không. Sử dụng các toán tử bitwise để xác định xem một số có phải là lũy thừa của 2
hay không.

```python
n = int(input())
if n > 0 and n & (n - 1) == 0:
    print("Day la luy thua cua 2")
else:
    print("Day khong phai luy thua cua 2")
```

W3A11 : Nhập vào 3 số nguyên dương a,b,c là độ dài 3 cạnh của một tam giác. Kiểm
tra 3 số đã nhập có tạo thành một tam giác hợp lệ hay không. Nếu không in
ra "Không phải tam giác". Nếu có, phân loại tam giác đó là "Tam giác đều",
"Tam giác cân", hay "Tam giác thường".

```python
a, b, c = map(int, input(). split())
if a+b>c and b+c>a and a+c>b : 
    if a == b == c : 
        print("Tam giac deu")
    elif a == b != c or a == c != b or b == c != a:
        print("Tam giac can")
    else : 
        print("Tam giac thuong")
else: 
    print("Khong phai tam giac")
```
