W4A6 : Nhập hai số nguyên dương a, b (a ≤ b). Hãy tính tổng các số nguyên tố trong đoạn
[a, b]. Mỗi dòng là một bộ test (a b)."

```python
try:
    while True:
        a, b = map(int, input().split())
        tong = 0
        for i in range(a, b + 1):
            if i >= 2:
                la_so_nguyen_to = True
                for j in range(2, int(i ** 0.5) + 1):
                    if i % j == 0:
                        la_so_nguyen_to = False
                        break
                if la_so_nguyen_to == True:
                    tong += i
        print(tong)
except (EOFError, ValueError):
    pass
```

W4A7 : Nhập vào số nguyên dương n (n ≥ 2). Hãy in ra ước số nguyên tố lớn nhất của n.

```python
n = int(input())
d = 2 
while n > 1: 
    if n%d == 0: 
        uoc_lon_nhat = d 
        while n%d == 0 : 
            n = n//d 
    d += 1
print(uoc_lon_nhat)
```

W4A8 : Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số
đối xứng (palindrome). In ra số bước và giá trị palindrome thu được. 

```python
n = int(input())
chuoi_n = str(n)
so_buoc = 0 
while chuoi_n != chuoi_n[::-1]:
    so_dao_nguoc = int(chuoi_n[::-1])
    n = n + so_dao_nguoc
    chuoi_n = str(n)
    so_buoc += 1 
print(so_buoc, n)
```

W4A9 : Nhập vào số nguyên dương n. In ra tất cả các số chính phương ≤ n có các chữ số
đôi một khác nhau (in theo thứ tự tăng dần, cùng hàng, cách nhau bởi một dấu
cách). Nếu không có, in ra (no number).

```python
n = int(input())
i = 0 
co_so_nao_khong = False 
while i*i <= n:
    so_chinh_phuong = i*i 
    chuoi_so = str(so_chinh_phuong)
    la_khac_nhau = True 
    for chu_so in chuoi_so:
        if chuoi_so.count(chu_so) > 1:
            la_khac_nhau = False
            break 
    if la_khac_nhau == True: 
        print(so_chinh_phuong, end =" ")
        co_so_nao_khong = True
    i += 1 
if co_so_nao_khong == False : 
    print("no number")
```

W9A10 : Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất
và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số
nhỏ nhất.

```python
n = int(input())
max_length = 0 
best_x = 1 
for i in range (1,n+1):
    current = i 
    length = 1 
    while current > 1: 
        if current % 2 == 0 : 
            current = current // 2
        else :
            current = current * 3 + 1
        length += 1 
    if length > max_length: 
        max_length = length 
        best_x = i 
print(best_x, max_length)
```

W4A11 : Viết chương trình đếm số lượng ước số chẵn của số nguyên dương n với n < 10^6.

Ví dụ 220 và 284 là một cặp số thân thiết
220 ngoài chính nó ra, nó còn có 11 ước số khác là: 1, 2, 4, 5, 10, 11, 20, 44, 55,
110. Tổng của 11 ước này cho chúng ta 284
284 ngoài bản thân nó, còn có 5 ước số khác là: 1, 2, 4, 71, 142. Tổng của chúng
vừa đúng bằng 220
Nhập hai số từ bàn phím và kiểm tra xem chúng có phải là cặp số thân thiết không.
Nếu có, in ra màn hình "true", ngược lại, in ra "false" 


```python
m, n = map(int, input(). split())
tong_uoc_m = 0 
for i in range (1, (m//2)+1):
    if m % i == 0 : 
        tong_uoc_m += i
tong_uoc_n = 0 
for i in range (1, (n//2)+1):
    if n % i == 0: 
        tong_uoc_n += i
if tong_uoc_m == n and tong_uoc_n == m: 
    print("true")
else: 
    print("false")
```
