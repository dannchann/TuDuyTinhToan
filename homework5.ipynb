W5A8 : Khoảng cách Hamming giữa hai số nguyên là số lượng vị trí khác nhau giữa hai dãy bits tương ứng
của chúng. Ví dụ: Khoảng cách Hamming giữa 1 (1) và 4 (100) là 2.

Cho hai số nguyên x và y, hãy viết hàm trả về khoảng cách Hamming giữa hai số này.
Khoảng cách Hamming là cái tên được đặt theo tên của Richard Hamming, người giới thiệu lý
thuyết này trong tài liệu có tính cơ sở của ông về mã phát hiện lỗi và sửa lỗi (error-detecting and
error-correcting codes). Nó được sử dụng trong kỹ thuật viễn thông để tính số lượng các bit trong
một từ nhị phân (binary word) bị đổi ngược, như một hình thức để ước tính số lỗi xảy ra trong quá
trình truyền thông, và vì thế, đôi khi, nó còn được gọi là khoảng cách tín hiệu (signal distance). Việc
phân tích trọng số Hamming của các bit còn được sử dụng trong một số ngành, bao gồm lý thuyết
tin học, lý thuyết mã hóa, và mật mã học. Tuy vậy, khi so sánh các dãy ký tự có chiều dài khác
nhau, hay các dãy ký tự có xu hướng không chỉ bị thay thế không thôi, mà còn bị ảnh hưởng bởi dữ
liệu bị lồng thêm vào, hoặc bị xóa đi, phương pháp đo lường phức tạp hơn, như khoảng cách
Levenshtein (Levenshtein distance) là một phương pháp có tác dụng và thích hợp hơn. 


```python
x, y = map(int, input(). split())
def khoang_cach_hamming(x, y): 
    ket_qua_xor = x^y 
    chuoi_nhi_phan = bin(ket_qua_xor)
    so_luong_khac_biet = chuoi_nhi_phan.count('1')
    return so_luong_khac_biet
print(khoang_cach_hamming(x, y))
```

được giữ không đổi. Hai kí tự khác nhau không thể ánh xạ đến cùng một kí tự, nhưng một kí tự có
thể ánh xạ thành chính nó. Ví dụ, từ "abca" và "zbxz" là hai từ đẳng cấu vì ta có thể ánh xạ 'a' thành
'z', 'b' thành 'b' và 'c' thành 'x'.

Viết hàm trả về \(true\) nếu hai từ \(a\) và \(b\) là đẳng cấu, ngược lại \(false\). Bạn có thể giả sử hai
từ đầu vào có cùng độ dài. 


```python
def kiem_tra_dang_cau(a, b): 
    if len(a) != len(b):
        return False 
    n = len(a)
    for i in range(n): 
        for j in range(i+1, n): 
            if a[i] == a[j] and b[i] != b[j]:
                return False 
            if a[i] != a[j] and b[i] == b[j]:
                return False 
    return True 
a = input()
b = input()
print(kiem_tra_dang_cau(a, b))
```

W5A16 : Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số
nguyên x

```python
def check_armstrong(x: int) -> bool:
    if x < 0:
        return False 
    str_x = str(x)
    n = len(str_x)
    total = sum(int(digit)**n for digit in str_x)
    return total == x 
x = int(input())
print(check_armstrong(x))
```

W5A25 : Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3),
(0,3)], p = (2,1)


```python
def point_in_quadrilateral(quad, p):
    def cross_product(a, b, pt):
        return (b[0] - a[0]) * (pt[1] - a[1]) - (b[1] - a[1]) * (pt[0] - a[0])
    def on_segment(a, b, pt):
        return (min(a[0], b[0]) <= pt[0] <= max(a[0], b[0]) and 
                min(a[1], b[1]) <= pt[1] <= max(a[1], b[1]))
    has_pos = False
    has_neg = False
    for i in range(4):
        p1 = quad[i]
        p2 = quad[(i + 1) % 4]
        cp = cross_product(p1, p2, p)
        if cp == 0:
            if on_segment(p1, p2, p):
                return True
        elif cp > 0:
            has_pos = True
        else:
            has_neg = True
        if has_pos and has_neg:
            return False
    return True
```

W5A28 : Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).


```python
import math 
def tinh_goc(x1, y1, x2, y2, x3, y3): 

    BA_x = x1 - x2
    BA_y = y1 - y2 

    BC_x = x3 - x2
    BC_y = y3 - y2 

    tich_vo_huong = (BA_x * BC_x) + (BA_y * BC_y)

    do_dai_BA = math.sqrt(BA_x**2 + BA_y**2)
    do_dai_BC = math.sqrt(BC_x**2 + BC_y**2)

    if do_dai_BA == 0 or do_dai_BC == 0: 
        return 0.0 
    
    goc = tich_vo_huong/(do_dai_BC*do_dai_BA) 
    goc = max(-1.0, min(1.0, goc))
    rad_goc = math.acos(goc)
    do_goc = math.degrees(rad_goc)
    return do_goc
x1, y1, x2, y2, x3, y3 = map(float, input().split())
print(tinh_goc(x1, y1, x2, y2, x3, y3))
```

W5A33 : Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa,
giá km, phí chờ, phụ phí ban đêm và giảm giá.

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type == 4: # Xe 4 chỗ
        gia_mo_cua = 10000      
        gia_km_tiep = 15000 
        phi_cho_1p = 500 
    elif car_type == 7:
        gia_mo_cua = 12000
        gia_km_tiep = 17000
        phi_cho_1p = 500
    else:
        return 0
    if km <= 1:
        tien_di_chuyen = gia_mo_cua
    else:
        tien_di_chuyen = gia_mo_cua + (km - 1) * gia_km_tiep
    tien_cho = wait_minutes * phi_cho_1p
    tong_tien = tien_di_chuyen + tien_cho
    if start_hour >= 22 or start_hour <= 5:
        tong_tien = tong_tien * 1.20 
    if km > 50:
        tong_tien = tong_tien * 0.90
    return round(tong_tien)
car_type, start_hour, km, wait_minutes = map(float, input().split())
print(smart_taxi_fare(car_type, start_hour, km, wait_minutes))
```

W5A34 : Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau
Nếu n ≤ 0 trả về 0.

```python
def sum_1_to_n(n): 
    if n <= 0: 
        return 0 
    
    total = 0 
    for x in range(1, n+1): 
        str_x = str(x)
        if len(str_x) == len(set(str_x)): 
            total += x 
    return total 
n = int(input())
print(sum_1_to_n(n))
```
